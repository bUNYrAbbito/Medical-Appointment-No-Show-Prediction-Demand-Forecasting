```python
import pandas as pd
import numpy as np

data = pd.read_csv("../Medical_appointment_data.csv")
```

```python
def clean_text(x):
    if pd.isna(x):
        return x
    return str(x).strip().lower()

cols = ['specialty','appointment_time','no_show','disability','place','appointment_shift']

for col in cols:
    data[col] = data[col].apply(clean_text)
```

```python
data['age'] = data['age'].fillna(data['age'].median())

cat_cols = ['specialty','disability','place','appointment_shift']
for col in cat_cols:
    data[col] = data[col].fillna('unknown')
```

```python
data.head()
```

```python
data['no_show'] = data['no_show'].map({'yes':1,'no':0})
data = data.dropna(subset=['no_show'])
```

```python
data
```

```python
data['appointment_date_continuous'] = pd.to_datetime(data['appointment_date_continuous'])
data['day'] = data['appointment_date_continuous'].dt.day
data['month'] = data['appointment_date_continuous'].dt.month
data['weekday'] = data['appointment_date_continuous'].dt.weekday

data['age'] = data['age'].astype(int)
data.drop('appointment_date_continuous', axis=1, inplace=True)
data
```

```python
data = data.ffill()
data = data.bfill()
data
```

```python
data = pd.get_dummies(data, drop_first=True)
data
```

```python
data.to_csv("../clean_data.csv", index=False)
```
